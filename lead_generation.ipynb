## Project Overview

This project demonstrates a complete lead generation automation pipeline using Python. The objective is to transform raw data into a structured and usable lead dataset that can be used for CRM and outreach purposes.

The workflow includes:
- Data loading
- Data transformation
- Email generation (bonus feature)
- Data cleaning
- Feature engineering
- Automated pipeline execution
- Export to Excel

```python
# Import required libraries
import pandas as pd
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
def run_pipeline():

    print("Starting Lead Generation Pipeline...")

    # STEP 1: Load dataset
    df = pd.read_csv("people-100.csv")
    print("Dataset loaded")

    # STEP 2: Simulate missing data
    df.loc[0:4, "Email"] = None

    # STEP 3: Create structured dataset
    df_clean = pd.DataFrame()

    df_clean["Name"] = df["First Name"] + " " + df["Last Name"]
    df_clean["Email"] = df["Email"]
    df_clean["Phone"] = df["Phone"]
    df_clean["Job_Title"] = df["Job Title"]

    # STEP 4: Email generation
    df_clean["Email"] = df_clean["Email"].fillna(
        df_clean["Name"].str.lower().str.replace(" ", "") + "@company.com"
    )

    df_clean["Email"] = df_clean["Email"].str.lower()

    # STEP 5: Enrichment
    df_clean["Website"] = "www.company.com"
    df_clean["Location"] = "Global"
    df_clean["Lead_Source"] = "Dataset"

    # STEP 6: Cleaning
    df_clean = df_clean.drop_duplicates()
    df_clean = df_clean.fillna("Not Available")

    # STEP 7: Features
    df_clean = df_clean.reset_index(drop=True)
    df_clean["Lead_ID"] = "L" + (df_clean.index + 1).astype(str)
    df_clean["Status"] = "New"
    df_clean["Priority"] = "Medium"

    # STEP 8: Final format
    df_clean = df_clean[
        ["Lead_ID", "Name", "Email", "Phone", "Job_Title",
         "Website", "Location", "Status", "Priority", "Lead_Source"]
    ]

    # STEP 9: Export
    df_clean.to_excel("lead_generation_output.xlsx", index=False)

    print("Pipeline executed successfully")

    return df_clean
```

```python
df_clean = run_pipeline()
```
